```python
import os
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

fig_dir = Path("report/figures/sales")
if not fig_dir.exists():
    for parent in Path.cwd().parents:
        candidate = parent / "report" / "figures" / "sales"
        if (parent / "pyproject.toml").exists() or (parent / "Taskfile.yml").exists():
            fig_dir = candidate
            break
fig_dir.mkdir(parents=True, exist_ok=True)

sales = pd.read_excel("sales.xlsx")
sales
```

```python
sales.dtypes
```

```python
from skimpy import skim

skim(sales)
```

```python
plt.figure(figsize=(8, 5))
ax = sns.countplot(data=sales, x="City", hue="City", legend=False)

for container in ax.containers:
    ax.bar_label(container)
plt.title("Transaction Volume by City")
plt.xlabel("City")
plt.ylabel("Transaction Count")
plt.savefig(fig_dir / "sales_transaction_volume_by_city.png", bbox_inches="tight", dpi=300)
plt.show()
```

Yanhon has the majority of sales by a small margin.

```python
plt.figure(figsize=(10, 5))
ax = sns.countplot(data=sales, x="City", hue="Product Code")

for container in ax.containers:
    ax.bar_label(container)
plt.title("Product Line Distribution Across Cities")
plt.xlabel("City")
plt.ylabel("Transaction Count")
plt.savefig(fig_dir / "sales_product_line_distribution_by_city.png", bbox_inches="tight", dpi=300)
plt.show()
```

Similar story per product, except Mandalay has a bigger amount of sales for product 3 compared to the other cities.

```python
plt.figure(figsize=(8, 5))
ax = sns.countplot(data=sales, x="Payment", hue="Payment", legend=False, palette="Set2")

for container in ax.containers:
    ax.bar_label(container)
plt.title("Transaction Frequency by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Transaction Count")
plt.savefig(fig_dir / "sales_payment_method_frequency.png", bbox_inches="tight", dpi=300)
plt.show()
```

Most sales are done with E-wallets and Credit cards, with ~30 sales difference compared to cash.

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
sns.boxplot(data=sales, x="Payment", hue="Payment", y="Unit price", palette="Set2", dodge=False, ax=ax1)
ax1.set_title("Unit Price Distribution (Boxplot)")
ax1.set_xlabel("Payment Method")
ax1.set_ylabel("Unit Price (USD)")

sns.kdeplot(
    data=sales,
    x="Unit price",
    hue="Payment",
    palette="Set2",
    common_norm=False,
    linewidth=2,
    ax=ax2
)
ax2.set_title("Unit Price Density Distribution")
ax2.set_xlabel("Unit Price (USD)")
ax2.set_ylabel("Density")

plt.tight_layout()
plt.savefig(fig_dir / "sales_unit_price_distribution_by_payment.png", bbox_inches="tight", dpi=300)
plt.show()
```
